```python
# ============================================================
# NOTEBOOK : FN_LOGGER
# PURPOSE  : CENTRALIZED LOGGING FUNCTIONS
# ============================================================

# ============================================================
# IMPORT PACKAGES
# ============================================================

from pyspark.sql.functions import *
from datetime import datetime

# ============================================================
# FUNCTION : INSERT AUDIT LOG
# ============================================================

def insert_audit_log(

    pipeline_run_id,
    pipeline_name,
    source_system,
    table_name,
    start_time,
    end_time,
    rows_read,
    rows_written,
    status,
    execution_time_seconds

):

    audit_df = spark.createDataFrame(

        [
            (
                pipeline_run_id,
                pipeline_name,
                source_system,
                table_name,
                start_time,
                end_time,
                rows_read,
                rows_written,
                status,
                execution_time_seconds
            )
        ],

        [
            "pipeline_run_id",
            "pipeline_name",
            "source_system",
            "table_name",
            "start_time",
            "end_time",
            "rows_read",
            "rows_written",
            "status",
            "execution_time_seconds"
        ]
    )

    audit_df.write.mode("append").saveAsTable(
        "config.audit_pipeline_log"
    )

    print(f"Audit Log Inserted : {table_name}")

# ============================================================
# FUNCTION : INSERT ERROR LOG
# ============================================================

def insert_error_log(

    error_id,
    pipeline_run_id,
    table_name,
    source_system,
    notebook_name,
    error_message

):

    error_df = spark.createDataFrame(

        [
            (
                error_id,
                pipeline_run_id,
                table_name,
                source_system,
                notebook_name,
                error_message,
                datetime.now()
            )
        ],

        [
            "error_id",
            "pipeline_run_id",
            "table_name",
            "source_system",
            "notebook_name",
            "error_message",
            "error_timestamp"
        ]
    )

    error_df.write.mode("append").saveAsTable(
        "config.error_log_table"
    )

    print(f"Error Log Inserted : {table_name}")

# ============================================================
# NOTEBOOK LOADED MESSAGE
# ============================================================

print("FN_LOGGER LOADED SUCCESSFULLY")
```
