```python
# ============================================================
# FUNCTION : GET METADATA
# ============================================================

def get_metadata(table_name):

    metadata_df = spark.sql(f"""

        SELECT *

        FROM config.metadata_table

        WHERE target_table = '{table_name}'

    """)

    metadata_row = metadata_df.collect()[0]

    metadata = {

        "source_system":
            metadata_row["source_system"],

        "source_folder":
            metadata_row["source_folder"],

        "file_name":
            metadata_row["file_name"],

        "target_schema":
            metadata_row["target_schema"],

        "target_table":
            metadata_row["target_table"],

        "load_type":
            metadata_row["load_type"],

        "primary_key_column":
            metadata_row["primary_key_column"],

        "watermark_column":
            metadata_row["watermark_column"]

    }

    return metadata

# FUNCTION : UPDATE WATERMARK
# ============================================================

def update_watermark(table_name,max_date):

    spark.sql(f"""

        UPDATE config.watermark_table

        SET

            last_load_timestamp =
                '{max_date}',

            updated_timestamp =
                current_timestamp()

        WHERE table_name = '{table_name}'

    """)

    print(f"Successfully the getted the Metadata information for the : {table_name}")

# ============================================================
```

```python
# FUNCTION : GET WATERMARK VALUE
# ============================================================

def get_watermark(table_name):

    watermark_df = spark.sql(f"""

        SELECT last_load_timestamp
        FROM config.watermark_table
        WHERE table_name = '{table_name}'

    """)

    watermark_value = watermark_df.collect()[0][0]

    return watermark_value

# ============================================================

# FUNCTION : UPDATE WATERMARK
# ============================================================

def update_watermark(table_name,max_date):

    spark.sql(f"""

        UPDATE config.watermark_table

        SET

            last_load_timestamp =
                '{max_date}',

            updated_timestamp =
                current_timestamp()

        WHERE table_name = '{table_name}'

    """)

    print(f"Watermark Updated : {table_name}")

# ============================================================
```

```python
# FUNCTION : GET MAX DATE
# ============================================================

def get_max_date(dataframe,column_name):

    max_date = dataframe.agg(
        max(column_name)
    ).collect()[0][0]

    return max_date

print("Max Date updated successfully")

# ============================================================
```
