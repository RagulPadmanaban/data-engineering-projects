#### GOLD SALES SUMMARY

```python
spark.sql("""

    CREATE SCHEMA IF NOT EXISTS gold

""")

print("Gold schema created")
```

```python
spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_sales_summary AS

    SELECT

        product_id,
        product_name,
        sales_category,

        SUM(quantity) AS total_quantity_sold,

        SUM(total_amount) AS total_sales_amount,

        AVG(unit_price) AS average_unit_price,

        COUNT(order_id) AS total_orders

    FROM silver.fact_order_items

    GROUP BY

        product_id,
        product_name,
        sales_category

""")

print("gold_sales_summary created successfully")
```

```python
display(

    spark.sql("""

        SELECT *
        FROM gold.gold_sales_summary

    """)

)
```

# **GOLD CUSTOMER ANALYTICS**

```python
# ============================================================
# GOLD CUSTOMER ANALYTICS
# ============================================================

spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_customer_analytics AS

    SELECT

        customer_id,

        customer_name,

        COUNT(DISTINCT order_id)
            AS total_orders,

        SUM(total_amount)
            AS total_spent,

        AVG(total_amount)
            AS average_order_value,

        MAX(modified_date)
            AS last_order_date,

        SUM(quantity)
            AS total_quantity_purchased,

        CASE

            WHEN SUM(total_amount) >= 100000

            THEN 'PREMIUM_CUSTOMER'

            WHEN SUM(total_amount) >= 50000

            THEN 'REGULAR_CUSTOMER'

            ELSE 'STANDARD_CUSTOMER'

        END AS customer_category

    FROM silver.fact_order_items

    GROUP BY

        customer_id,
        customer_name

""")

print("gold_customer_analytics created successfully")
```

# **GOLD PRODUCT ANALYTICS**

```python
# ============================================================
# GOLD PRODUCT ANALYTICS
# ============================================================

spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_product_analytics AS

    SELECT

        product_id,

        product_name,

        SUM(quantity)
            AS total_quantity_sold,

        SUM(total_amount)
            AS total_revenue,

        AVG(unit_price)
            AS average_unit_price,

        COUNT(DISTINCT order_id)
            AS total_orders,

        MAX(modified_date)
            AS last_sale_date,

        CASE

            WHEN SUM(total_amount) >= 100000

            THEN 'HIGH_PERFORMING_PRODUCT'

            WHEN SUM(total_amount) >= 50000

            THEN 'MEDIUM_PERFORMING_PRODUCT'

            ELSE 'LOW_PERFORMING_PRODUCT'

        END AS product_performance_category

    FROM silver.fact_order_items

    GROUP BY

        product_id,
        product_name

""")

print("gold_product_analytics created successfully")
```

# **GOLD PAYMENT ANALYTICS**

```python
# ============================================================
# GOLD PAYMENT ANALYTICS
# ============================================================

spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_payment_analytics AS

    SELECT

        payment_method,

        payment_status,

        COUNT(payment_id)
            AS total_payments,

        SUM(payment_amount)
            AS total_payment_amount,

        AVG(payment_amount)
            AS average_payment_amount,

        MAX(payment_date)
            AS last_payment_date,

        CASE

            WHEN SUM(payment_amount) >= 100000

            THEN 'HIGH_PAYMENT_VOLUME'

            WHEN SUM(payment_amount) >= 50000

            THEN 'MEDIUM_PAYMENT_VOLUME'

            ELSE 'LOW_PAYMENT_VOLUME'

        END AS payment_volume_category

    FROM silver.fact_payments

    GROUP BY

        payment_method,
        payment_status

""")

print("gold_payment_analytics created successfully")
```

# **GOLD INVENTORY SUMMARY**

```python
# ============================================================
# GOLD INVENTORY SUMMARY
# ============================================================

spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_inventory_summary AS

    SELECT

        product_id,

        product_name,

        warehouse_name,

        stock_quantity,

        reorder_level,

        last_restock_date,

        CASE

            WHEN stock_quantity <= reorder_level

            THEN 'REORDER_REQUIRED'

            WHEN stock_quantity <= (reorder_level * 2)

            THEN 'LOW_STOCK'

            ELSE 'STOCK_AVAILABLE'

        END AS inventory_status

    FROM silver.dim_inventory

    WHERE is_current = 1
      AND is_deleted = 0

""")

print("gold_inventory_summary created successfully")
```

# **GOLD SALES TREND ANALYTICS**

```python
# ============================================================
# GOLD SALES TREND ANALYTICS
# ============================================================

spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_sales_trend_analytics AS

    SELECT

        CAST(order_date AS DATE)
            AS sales_date,

        sales_category,

        COUNT(DISTINCT order_id)
            AS total_orders,

        SUM(quantity)
            AS total_quantity_sold,

        SUM(total_amount)
            AS total_sales_amount,

        AVG(total_amount)
            AS average_sales_amount

    FROM silver.fact_order_items

    GROUP BY

        CAST(order_date AS DATE),
        sales_category

""")

print("gold_sales_trend_analytics created successfully")
```

# **GOLD CATEGORY SALES ANALYTICS**

```python
# ============================================================
# GOLD CATEGORY SALES ANALYTICS
# ============================================================

spark.sql("""

    CREATE OR REPLACE TABLE gold.gold_category_sales_analytics AS

    SELECT

        category_name,

        COUNT(DISTINCT order_id)
            AS total_orders,

        SUM(quantity)
            AS total_quantity_sold,

        SUM(total_amount)
            AS total_sales_amount,

        AVG(total_amount)
            AS average_sales_amount,

        MAX(modified_date)
            AS last_sale_date,

        CASE

            WHEN SUM(total_amount) >= 100000

            THEN 'HIGH_SALES_CATEGORY'

            WHEN SUM(total_amount) >= 50000

            THEN 'MEDIUM_SALES_CATEGORY'

            ELSE 'LOW_SALES_CATEGORY'

        END AS category_performance

    FROM silver.fact_order_items

    GROUP BY

        category_name

""")

print("gold_category_sales_analytics created successfully")
```
